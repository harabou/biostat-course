```python
from ultralytics import YOLO

model_path = 'best.pt'
photo_dir = 'test_photos/'

model = YOLO(model_path)
```

## 3. 検出してみる

```python
import os

# 使える写真の一覧を確認
print(os.listdir(photo_dir))
```

```python
import matplotlib.pyplot as plt

def detect_and_show(image_path, confidence=0.4):
    result = model.predict(image_path, conf=confidence, verbose=False)[0]

    annotated = result.plot()       # 検出結果を描き込んだ画像(BGR)
    annotated_rgb = annotated[..., ::-1]  # BGR -> RGB

    fig, ax = plt.subplots(1, figsize=(8, 8))
    ax.imshow(annotated_rgb)
    ax.axis('off')
    plt.show()
    return result
```

```python
# 配布された写真の1枚を試してみる(ファイル名は実際のものに書き換えてください)
result = detect_and_show(photo_dir + "karaage2.png")
```

## 4. 結果を見てみよう

`result`の中身を見ると、検出された品目ごとに「信頼度スコア(confidence)」が入っています。
スコアが低い品目、複数の品目が重なって検出されている箇所などに注目してみましょう。

```python
for box in result.boxes:
    class_id = int(box.cls[0])
    class_name = model.names[class_id]
    confidence = float(box.conf[0])
    print(f"{class_name}: confidence = {confidence:.2f}")
```
